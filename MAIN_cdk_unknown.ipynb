```python
from padelpy import padeldescriptor

fingerprint = 'CDK'

fingerprint_output_file = ''.join([fingerprint,'.csv']) #CDK.csv
fingerprint_descriptortypes = fp[fingerprint]

padeldescriptor(mol_dir='molecule.smi',
                d_file=fingerprint_output_file, #'CDK.csv'
                #descriptortypes='CDKFingerprintCount.xml',
                descriptortypes= fingerprint_descriptortypes,
                detectaromaticity=True,
                standardizenitro=True,
                standardizetautomers=True,
                threads=1,
                removesalt=True,
                log=True,
                fingerprints=True)
```

```python
descriptors = pd.read_csv(fingerprint_output_file)
descriptors
```

```python
# Load the original data to get 'imppat id'
original_df = pd.read_excel('/content/New Microsoft Excel Worksheet (2).xlsx')

# Load the generated descriptor file
descriptors_df = pd.read_csv(fingerprint_output_file)

# Concatenate 'imppat id' from the original data with descriptors
final_df = pd.concat([original_df[['IMPPAT_id']], descriptors_df], axis=1)

# Save the final output with 'imppat id'
final_output_file = f"{fingerprint}_with_imppat_id.csv"
final_df.to_csv(final_output_file, index=False)

print(f"Final file saved as {final_output_file}")
```
